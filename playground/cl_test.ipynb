```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
lens = torch.tensor([3, 4, 2, 5])  # Example sequence lengths for a batch of 4 samples
y_true = torch.arange(len(lens))  # Create y_true as [0, 1, 2, 3]
print("y_true:", y_true)
```

```python
y_i = y_true.unsqueeze(1) # (B, 1)
y_j = y_true.unsqueeze(0) # (1, B)
print("y_i:", y_i)
print("y_j:", y_j)
```

```python
y_ij = y_i + y_j  # (B, B)
print("y_ij:", y_ij)
```
