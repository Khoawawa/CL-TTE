```python
def do_stat(data):
    y = data[:, -1].copy()
    trip = data[:, 1].copy()
    trip_len = np.array([len(trip[i]) for i in range(len(trip))])
    
    print("=== Segment Count Distribution ===")
    for p in [10, 25, 50, 75, 90, 95, 99, 100]:
        print(f"  p{p:02d}: {np.percentile(trip_len, p):.0f} segments")

    print("\n=== Travel Time Distribution (seconds) ===")
    for p in [10, 25, 50, 75, 90, 95, 99]:
        t = np.percentile(y, p)
        print(f"  p{p:02d}: {t:.0f}s  ({t/60:.1f} min)")

    print("\n=== Seconds Per Segment ===")
    sps = y / trip_len
    for p in [10, 25, 50, 75, 90, 95, 99]:
        print(f"  p{p:02d}: {np.percentile(sps, p):.1f} s/segment")

    print("\n=== Key Stats ===")
    print(f"median segments : {np.median(trip_len):.0f}")
    print(f"median duration : {np.median(y):.0f}s  ({np.median(y)/60:.1f} min)")
    print(f"mean segments   : {trip_len.mean():.0f}")
    print(f"mean duration   : {y.mean():.0f}s")
    print(f"total trips     : {len(y):,}")
    
    return trip_len, y
```
